```python
import sys, os
sys.path.append(os.path.abspath('..'))  # so `import src` works from notebooks/
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from src import config
from src.data.download_data import download
from src.data.preprocess import run as preprocess_run
sns.set_theme(style='whitegrid')
```

## Acquire + clean
Tries the real UCI data; falls back to the offline fixture.

```python
download()                 # real data if online, else synthetic fixture
train_df, test_df = preprocess_run()
df = pd.read_csv(config.CLEAN_DATA_PATH)
df.head()
```

## Shape, dtypes, missing values

```python
print(df.shape)
display(df.describe().T)
print('missing values:\n', df.isna().sum())
```

## Class balance

```python
ax = df[config.TARGET].value_counts().sort_index().plot.bar()
ax.set(title='Class balance', xlabel='target', ylabel='count'); plt.show()
```

## Numeric distributions by target

```python
for col in config.NUMERIC_FEATURES:
    sns.histplot(data=df, x=col, hue=config.TARGET, kde=True, alpha=0.5)
    plt.title(col); plt.show()
```

## Correlation heatmap

```python
plt.figure(figsize=(9,7))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation heatmap'); plt.show()
```
